```python
import pandas as pd

results = pd.read_csv('../data/results.csv')
results.head()
```

```python
print(results.shape)
print(results.columns.tolist())
```

```python
races = pd.read_csv('../data/races.csv')
drivers = pd.read_csv('../data/drivers.csv')

races.head()
```

```python
drivers.head()
```

```python
results['top10_finish'] = (results['positionOrder'] <= 10).astype(int)
results[['driverId', 'raceId', 'grid', 'positionOrder', 'top10_finish']].head(10)
```

```python
results[['driverId', 'raceId', 'grid', 'positionOrder', 'top10_finish']].tail(10)
```

```python
df =  results.merge(races[['raceId', 'year', 'name', 'circuitId']], on='raceId')
df = df.merge(drivers[['driverId', 'forename', 'surname', 'nationality']], on='driverId')

df[['year', 'name', 'forename', 'surname', 'grid', 'positionOrder', 'top10_finish']].head(10)
```

```python
df = df.sort_values(['driverId' , 'year' , 'raceId'])
df['avg_last5_finish'] = (
    df.groupby('driverId')['positionOrder']
    .transform(lambda x: x.shift(1).rolling(5, min_periods=1).mean())
)

df[['forename', 'surname', 'year', 'name', 'positionOrder', 'avg_last5_finish']].head(15)
```

```python
df['constructor_avg_last5'] = (
    df.groupby('constructorId')['positionOrder']
    .transform(lambda x: x.shift(1).rolling(5, min_periods=1).mean())
)

df[['forename', 'surname', 'name', 'year', 'positionOrder', 'avg_last5_finish', 'constructor_avg_last5']].head(15)
```

```python
df['grid_vs_form'] = df['avg_last5_finish'] - df['grid']
```

```python
df[df['constructorId'] == df.loc[370, 'constructorId']][['forename', 'surname', 'year', 'name', 'positionOrder']].head(20)
```

```python
constructor_sorted = df.sort_values(['constructorId', 'year', 'raceId'])
constructor_sorted['constructor_avg_last5'] = (
    constructor_sorted.groupby('constructorId')['positionOrder']
    .transform(lambda x: x.shift(1).rolling(5, min_periods=1).mean())
)
df['constructor_avg_last5'] = constructor_sorted['constructor_avg_last5']
```

```python
df[(df['constructorId'] == df.loc[370, 'constructorId']) & (df['year'] == 2007)][['forename', 'surname', 'name', 'positionOrder']]
```

```python
df[['forename', 'surname', 'name', 'year', 'positionOrder', 'avg_last5_finish', 'constructor_avg_last5']].head(15)
```

```python
df_model =  df.dropna(subset= ['avg_last5_finish','constructor_avg_last5','grid_vs_form'])
print(df.shape,'->',df_model.shape)
```

```python
features = ['grid','avg_last5_finish','constructor_avg_last5','grid_vs_form']
train = df_model[df_model['year'] <=2019]
test =  df_model[df_model['year'] >2019]
X_train , y_train = train[features],train['top10_finish']
X_test , y_test = test[features],test['top10_finish']
print(X_train.shape, X_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression 
model = LogisticRegression()
model.fit(X_train,y_train)
print("Train accuracy:",model.score(X_train,y_train))
print("Test accuracy:",model.score(X_test,y_test))
```

```python
baseline_model = LogisticRegression()
baseline_model.fit(train[['grid']],y_train)
print("Baseline (grid only) test accuracy:",baseline_model.score(test[['grid']],y_test))
```

```python
from sklearn.metrics import classification_report , confusion_matrix
y_pred = model.predict(X_test)
print(confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

```python
from sklearn.metrics import roc_auc_score
y_prob = model.predict_proba(X_test)[:,1]
roc_auc = roc_auc_score(y_test,y_prob)
print("ROC-AUC:",roc_auc)
```

```python
import pandas as pd 
coef_df = pd.DataFrame({
    'feature':features,
    'coefficient':model.coef_[0]})
print(coef_df)
```

```python
sample = test.copy()
sample['predicted_prob']  = model.predict_proba(X_test)[:,1]
sample['predicted_top10'] = model.predict(X_test)

sample[['forename', 'surname', 'name', 'year', 'positionOrder', 'top10_finish','predicted_prob','predicted_top10']].sort_values('year').head(20)
```

```python
sample[['forename', 'surname', 'name', 'year', 'positionOrder', 'top10_finish','predicted_prob','predicted_top10']].sort_values('predicted_prob',ascending = False).head(20)
```

```python
from sklearn.ensemble import RandomForestClassifier
rf_model = RandomForestClassifier(
    n_estimators = 100,
    random_state = 42  
)
rf_model.fit(X_train,y_train)
print('RF Train accuracy:' , rf_model.score(X_train,y_train))
print('RF Test accuracy:' , rf_model.score(X_test,y_test))
rf_proba = rf_model.predict_proba(X_test)[:,1]
print('RF ROC-AUC:',roc_auc_score(y_test,rf_proba))
```

```python
qualifying = pd.read_csv('../data/qualifying.csv')
print(qualifying.shape)
print(qualifying.columns.tolist())
qualifying.head()
```

```python
def time_to_seconds(time):
    if pd.isna(time):
        return None
    time = str(time).strip()

    if ':' not in time:
        return None

    minutes, seconds = time.split(':', 1)
    return int(minutes) * 60 + float(seconds)
```

```python
print(time_to_seconds('1:26.572'))
```

```python
for col in ['q1','q2','q3']:
    qualifying[col + '_seconds'] = qualifying[col].apply(time_to_seconds)
```

```python
qualifying[['qualifyId', 'raceId', 'driverId', 'constructorId', 'number', 'position', 'q1', 'q2', 'q3','q1_seconds', 'q2_seconds', 'q3_seconds']].head()
```

```python
qualifying['best_quali_time'] =qualifying[['q1_seconds', 'q2_seconds', 'q3_seconds']].min(axis=1)
```

```python
pole_times = (
    qualifying.groupby('raceId')['best_quali_time']
    .min()
    .reset_index()
    .rename(columns={'best_quali_time': 'pole_time'})
)
    
```

```python
pole_time_map = pole_times.set_index('raceId')['pole_time']

qualifying['pole_time'] = qualifying['raceId'].map(pole_time_map)
```

```python
qualifying['qualifying_gap'] = (
    qualifying['best_quali_time']
    - qualifying['pole_time']
)
```
